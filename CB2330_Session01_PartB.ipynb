## Part B. Trying more Formulas - Geometric Mean

The geometric mean is a type of average for quantities that multiply, such as fold changes. It multiplies a finite set of positive real numbers together and then takes the $n^{th}$ root of the product, where $n$ is the total count of numbers.

$$\left(\prod_{i=1}^{N} x_i\right)^{1/N}$$


```python
numbers = [7, 5, 3, 4]
assert all(x > 0 for x in numbers)

# as a loop:
def geom_mean(numbers):
    product = 1
    for i in numbers:
      product *= i
    product = product ** (1/len(numbers))
    return product

result_loop = geom_mean(numbers)

# as one line in numpy:
import numpy as np

result_numpy = (np.prod(numbers))**(1/len(numbers))

assert result_loop == result_numpy
print(result_loop, "==", result_numpy, "yay!")
```
